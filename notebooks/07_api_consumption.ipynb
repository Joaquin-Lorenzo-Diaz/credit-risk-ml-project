```python
import requests

# Public API with the exchange rates for Argentina (official, blue, MEP, etc.)
url = "https://dolarapi.com/v1/dolares"

response = requests.get(url, timeout=15)

print(f"Status code: {response.status_code}")
print(response.json())
```

```python
import pandas as pd

# The API returns a list of dictionaries (one per rate type): pandas turns it into a table
df_usd = pd.DataFrame(response.json())
df_usd
```

```python
official = df_usd[df_usd['casa'] == 'oficial']['venta'].values[0]
blue = df_usd[df_usd['casa'] == 'blue']['venta'].values[0]

# FX gap: how much more expensive the blue dollar is than the official one, in %
gap = (blue - official) / official * 100

print(f"Official USD (sell): ${official:,.2f}")
print(f"Blue USD (sell): ${blue:,.2f}")
print(f"FX gap: {gap:.2f}%")
```

```python
# Business rule: turn the number into a decision.
# The thresholds (10% / 30%) are illustrative, not calibrated on historical data.
def evaluate_gap(gap):
    if gap < 10:
        return "NORMAL"
    elif gap < 30:
        return "WARNING"
    else:
        return "ALERT"

status = evaluate_gap(gap)
print(f"FX gap: {gap:.2f}% → Status: {status}")
```
