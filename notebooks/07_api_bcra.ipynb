```python
import requests

cuit = "30546689979"  # YPF S.A.
url = f"https://api.bcra.gob.ar/centraldedeudores/v1.0/Deudas/{cuit}"

response = requests.get(url, verify=False)

print(f"Código de respuesta: {response.status_code}")
print(response.json())
```

```python
import requests

url = "https://dolarapi.com/v1/dolares"

response = requests.get(url, timeout=15)

print(f"Código de respuesta: {response.status_code}")
print(response.json())
```

```python
import pandas as pd

df_dolar = pd.DataFrame(response.json())
df_dolar
```

```python
oficial = df_dolar[df_dolar['casa'] == 'oficial']['venta'].values[0]
blue = df_dolar[df_dolar['casa'] == 'blue']['venta'].values[0]

brecha = (blue - oficial) / oficial * 100

print(f"Dólar oficial (venta): ${oficial:,.2f}")
print(f"Dólar blue (venta): ${blue:,.2f}")
print(f"Brecha cambiaria: {brecha:.2f}%")
```

```python
def evaluar_brecha(brecha):
    if brecha < 10:
        return "NORMAL"
    elif brecha < 30:
        return "ATENCIÓN"
    else:
        return "ALERTA"

estado = evaluar_brecha(brecha)
print(f"Brecha cambiaria: {brecha:.2f}% → Estado: {estado}")
```
